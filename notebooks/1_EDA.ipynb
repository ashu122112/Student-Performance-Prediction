# Student Performance Indicator - Exploratory Data Analysis (EDA)

## 1. Objective
Understand how student performance (specifically `math_score`) is affected by various demographic and educational background variables such as Gender, Ethnicity, Parental Level of Education, Lunch type, and Test Preparation Course.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
warnings.filterwarnings('ignore')
```

## 2. Load Dataset

```python
df = pd.read_csv('../data/stud.csv')
df.head()
```

## 3. Data Checks & Summary
- Check missing values
- Check duplicate records
- Check data types & dataset shape

```python
print(f"Dataset Shape: {df.shape}")
print("\n--- Dataset Info ---")
df.info()
```

```python
print("Missing Values:\n", df.isna().sum())
print("\nDuplicate Count:", df.duplicated().sum())
print("\nUnique Values Per Column:\n", df.nunique())
```

```python
df.describe()
```

## 4. Feature Extraction & Categorization

```python
# Separate Numerical and Categorical columns
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

print(f"Numerical Features ({len(numeric_features)}): {numeric_features}")
print(f"Categorical Features ({len(categorical_features)}): {categorical_features}")
```

```python
# Adding Total Score and Average Score features for deeper EDA
df['total score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average'] = df['total score'] / 3
df.head()
```

## 5. Visual Data Analysis

```python
# Visualizing distributions of Math, Reading, and Writing Scores
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
plt.subplot(131)
sns.histplot(data=df, x='math_score', kde=True, color='purple')
plt.title('Math Score Distribution')

plt.subplot(132)
sns.histplot(data=df, x='reading_score', kde=True, color='teal')
plt.title('Reading Score Distribution')

plt.subplot(133)
sns.histplot(data=df, x='writing_score', kde=True, color='coral')
plt.title('Writing Score Distribution')
plt.show()
```

```python
# Impact of Test Preparation Course on Math Score
plt.figure(figsize=(8, 5))
sns.boxplot(x='test_preparation_course', y='math_score', data=df, palette='Set2')
plt.title('Math Score vs Test Preparation Course')
plt.show()
```

```python
# Impact of Lunch type on Performance
plt.figure(figsize=(8, 5))
sns.barplot(x='lunch', y='math_score', hue='gender', data=df, palette='viridis')
plt.title('Math Score by Lunch Type and Gender')
plt.show()
```
